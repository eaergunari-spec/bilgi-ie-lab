# Code Moment 1 — Can the computer beat your intuition?

You have **20 couriers** and four Istanbul zones. The forecast and courier productivity are fixed.

**Goal:** maximize the number of orders served.

Run the cell below. Then ask yourself: **Is the mathematically optimal plan automatically a good decision?**

```python
forecast = [32, 28, 24, 16]
rate = [4, 3, 5, 2]
zones = ['Beyoglu', 'Sisli', 'Kagithane', 'Besiktas']

best_score = -1
best_plan = None

for a in range(21):
    for b in range(21-a):
        for c in range(21-a-b):
            d = 20-a-b-c
            plan = [a, b, c, d]

            served = sum(
                min(forecast[i], rate[i] * plan[i])
                for i in range(4)
            )

            if served > best_score:
                best_score = served
                best_plan = plan

print('Best allocation:')
for zone, couriers in zip(zones, best_plan):
    print(f'  {zone:10s}: {couriers} couriers')
print('\nOrders served:', best_score)
```

## Think before moving on

The optimizer did exactly what we asked: **maximize total served orders**.

But if one zone receives no couriers, did we define the decision problem well enough?

> **A model does not know what you meant. It only knows what you wrote.**

```python
# Optional experiment: require every zone to serve at least 50% of forecast demand.
best_score = -1
best_plan = None

for a in range(21):
    for b in range(21-a):
        for c in range(21-a-b):
            d = 20-a-b-c
            plan = [a, b, c, d]
            served_by_zone = [min(forecast[i], rate[i]*plan[i]) for i in range(4)]
            service_levels = [served_by_zone[i]/forecast[i] for i in range(4)]

            if min(service_levels) >= 0.50:
                total = sum(served_by_zone)
                if total > best_score:
                    best_score = total
                    best_plan = plan

print('Best plan with a 50% minimum service level:', best_plan)
print('Orders served:', best_score)
```
